```python
# imports
import os
import json
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from scraper import fetch_website_links, fetch_website_contents
from openai import OpenAI
from torch.cpu import stream
```

```python
# constants

MODEL_GPT = 'gpt-4o-mini'
MODEL_LLAMA = 'llama3.2:latest'
```

```python
# setup environment
load_dotenv()
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY"))

OLLAMA_BASE_URL = os.getenv("OLLAMA_BASE_URL", "http://localhost:11434/v1")
ollama = OpenAI(base_url=OLLAMA_BASE_URL, api_key='pepsi')

openai = OpenAI()
```

```python
# here is the question; type over this to ask something new

question = """
Please explain what this code does and why:

main( ) {
        printf("hello, world");
}
"""
```

```python
# Get gpt-4o-mini to answer, with streaming



def get_answer(question):
    stream = openai.chat.completions.create(
        model=MODEL_GPT,
        stream=True,
        messages=[
            {"role": "system", "content": "You are a helpful assistant that explains code. It is important to be concise and clear."
            "Explanation must be in markdown format, and include code snippets if necessary."
            "Explanation need to be in a level that a beginner programmer can understand. By beginnig programmer, I mean someone who hasn't code before, but is eager to learn."
            "Explanation should be in Finnish"
            "Analyse code what is good and what is bad, and explain why. If there are any errors in the code, explain what they are and how to fix them. This section must not bee too long and should be own section in markdown format."},
            {"role": "user", "content": question}
        ]
    )
    answer = ""
    display_handle = display(Markdown(""), display_id=True)
    for chunk in stream:
        answer += chunk.choices[0].delta.content or ''
        update_display(Markdown(answer), display_id=display_handle.display_id)


get_answer(question)
```
